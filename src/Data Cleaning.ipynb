# Adult Income Data Cleaning and EDA

```python
# import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests

from pathlib import Path
```

```python
# define project paths

PROJECT_DIR = Path("..")

DATA_DIR = PROJECT_DIR / "data"

REPORTS_DIR = PROJECT_DIR / "reports"

CHARTS_DIR = REPORTS_DIR / "charts"

TABLES_DIR = REPORTS_DIR / "summary_tables"

print("Data folder:", DATA_DIR)
print("Reports folder:", REPORTS_DIR)
```

```python
# define dataset urls

adult_data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

adult_test_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test"
```

```python
# define file paths

adult_data_path = DATA_DIR / "adult.data"

adult_test_path = DATA_DIR / "adult.test"

print(adult_data_path)
print(adult_test_path)
```

```python
# download adult.data

response = requests.get(adult_data_url)

if response.status_code == 200:
    adult_data_path.write_bytes(response.content)
    print("adult.data downloaded successfully")
else:
    print("Failed to download adult.data")
    print("Status code:", response.status_code)
```

```python
# download adult.test

response = requests.get(adult_test_url)

if response.status_code == 200:
    adult_test_path.write_bytes(response.content)
    print("adult.test downloaded successfully")
else:
    print("Failed to download adult.test")
    print("Status code:", response.status_code)
```

```python
# verify downloaded files

print("adult.data exists:", adult_data_path.exists())
print("adult.test exists:", adult_test_path.exists())
```

```python
# define column names

columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]

print("Total Columns:", len(columns))
print(columns)
```

```python
# load adult.data

adult_data = pd.read_csv(
    adult_data_path,
    header=None,
    names=columns
)

adult_data.head()
```

```python
# load adult.test

adult_test = pd.read_csv(
    adult_test_path,
    header=None,
    names=columns,
    skiprows=1
)

adult_test.head()
```

```python
# check dataset shapes

print("adult.data shape :", adult_data.shape)
print("adult.test shape :", adult_test.shape)
```

```python
# inspect raw datasets

print("adult.data info")
adult_data.info()

print("\nadult.test info")
adult_test.info()
```

```python
# display sample records

adult_data.sample(5, random_state=42)
```

```python
# add source split

adult_data["source_split"] = "train"
adult_test["source_split"] = "test"
```

```python
# combine train and test data

df = pd.concat(
    [adult_data, adult_test],
    ignore_index=True
)

print("Combined dataset shape:", df.shape)
```

```python
# inspect combined dataset

df.head()
```

```python
# check last records

df.tail()
```

```python
# check dataset shape

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# check source split

print(df["source_split"].value_counts())
```

```python
# check source split percentage

print(
    df["source_split"].value_counts(normalize=True) * 100
)
```

```python
# check column names

print(df.columns.tolist())
```

```python
# save combined raw dataset

raw_combined_path = DATA_DIR / "adult_income_combined_raw.csv"

df.to_csv(raw_combined_path, index=False)

print("Combined raw dataset saved successfully.")
```

```python
# check categorical columns

categorical_columns = df.select_dtypes(include="object").columns

print(categorical_columns)
```

```python
# remove leading and trailing whitespace

for column in categorical_columns:
    df[column] = df[column].str.strip()
```

```python
# check question mark values

question_mark_counts = (df == "?").sum()

print(question_mark_counts)
```

```python
# check empty string values

empty_counts = (df == "").sum()

print(empty_counts)
```

```python
# convert question marks and empty strings to missing values

df = df.replace(["?", ""], np.nan)
```

```python
# check missing values

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# calculate missing value percentage

missing_percentage = (
    df.isnull().sum() / len(df) * 100
)

print(missing_percentage)
```

```python
# display missing value summary

missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": df.isnull().sum() / len(df) * 100
})

missing_summary = missing_summary[
    missing_summary["missing_count"] > 0
]

missing_summary
```

```python
# define categorical columns with missing values

missing_categorical_columns = [
    "workclass",
    "occupation",
    "native_country"
]
```

```python
# replace missing categorical values with Unknown

for column in missing_categorical_columns:
    df[column] = df[column].fillna("Unknown")
```

```python
# verify categorical missing values

print(df[missing_categorical_columns].isnull().sum())
```

```python
# count Unknown values

for column in missing_categorical_columns:
    print(column, ":", (df[column] == "Unknown").sum())
```

```python
# check remaining missing values

print(df.isnull().sum())
```

```python
# define numeric columns

numeric_columns = [
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

print(numeric_columns)
```

```python
# convert numeric columns

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
# check numeric missing values

print(df[numeric_columns].isnull().sum())
```

```python
# inspect numeric columns

df[numeric_columns].describe()
```

```python
# check impossible numeric values

print("Age below 0:", (df["age"] < 0).sum())
print("Education number below 0:", (df["education_num"] < 0).sum())
print("Hours below 0:", (df["hours_per_week"] < 0).sum())
print("Capital gain below 0:", (df["capital_gain"] < 0).sum())
print("Capital loss below 0:", (df["capital_loss"] < 0).sum())
```

```python
# check invalid core numeric values

print("Missing age:", df["age"].isnull().sum())
print("Missing education_num:", df["education_num"].isnull().sum())
print("Missing hours_per_week:", df["hours_per_week"].isnull().sum())

print("Age below 1:", (df["age"] < 1).sum())
print("Hours per week below 1:", (df["hours_per_week"] < 1).sum())
```

```python
# check numeric data types

print(df[numeric_columns].dtypes)
```

```python
# remove rows with invalid core numeric values

before_numeric_cleaning = len(df)

valid_numeric_mask = (
    df["age"].notna()
    & (df["age"] > 0)
    & df["education_num"].notna()
    & (df["education_num"] > 0)
    & df["hours_per_week"].notna()
    & (df["hours_per_week"] > 0)
)

df = df[valid_numeric_mask].copy()

after_numeric_cleaning = len(df)

print("Rows before numeric cleaning:", before_numeric_cleaning)
print("Rows after numeric cleaning:", after_numeric_cleaning)
print("Rows removed:", before_numeric_cleaning - after_numeric_cleaning)
```

```python
# check duplicate rows

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
# calculate duplicate percentage

duplicate_percentage = (
    duplicate_count / len(df) * 100
)

print("Duplicate percentage:", duplicate_percentage)
```

```python
df.drop_duplicates()
```

```python
# check income labels

print(df["income"].value_counts())
```

```python
# remove trailing period from income labels

df["income"] = df["income"].str.rstrip(".")
```

```python
# verify income labels

print(df["income"].value_counts())
```

```python
# check unique income labels

print(df["income"].unique())
```

```python
# validate income labels

valid_income_labels = ["<=50K", ">50K"]

invalid_income_labels = ~df["income"].isin(valid_income_labels)

print("Invalid income labels:", invalid_income_labels.sum())
```

```python
# create high income flag

df["high_income"] = (
    df["income"] == ">50K"
).astype(int)
```

```python
# check high income flag

print(df["high_income"].value_counts())
```

```python
# compare income and high income

print(
    pd.crosstab(
        df["income"],
        df["high_income"]
    )
)
```

```python
# calculate high income rate

high_income_rate = df["high_income"].mean() * 100

print("Above $50K rate:", high_income_rate)
```

```python
# create age groups

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 45, 55, 65, np.inf],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ]
)
```

```python
# check age groups

print(df["age_group"].value_counts().sort_index())
```

```python
# create weekly working hour groups

df["hours_band"] = pd.cut(
    df["hours_per_week"],
    bins=[0, 20, 40, 60, np.inf],
    labels=[
        "<=20",
        "21-40",
        "41-60",
        "61+"
    ]
)
```

```python
# check weekly hour groups

print(df["hours_band"].value_counts().sort_index())
```

```python
# create net capital gain or loss

df["net_capital"] = (
    df["capital_gain"] - df["capital_loss"]
)
```

```python
# check newly created features

print(
    df[
        [
            "age_group",
            "hours_band",
            "net_capital",
            "high_income"
        ]
    ].head()
)
```

```python
# check final dataset shape

print("Final rows:", df.shape[0])
print("Final columns:", df.shape[1])
```

```python
# final missing value check

missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": df.isnull().sum() / len(df) * 100
})

missing_summary[missing_summary["missing_count"] > 0]
```

```python
# check final data types

print(df.dtypes)
```

```python
# check final columns

print(df.columns.tolist())
```

```python
# validate income values

print("Unique income values:")
print(df["income"].unique())

print("\nHigh income values:")
print(df["high_income"].unique())
```

```python
# validate high income mapping

income_check = pd.crosstab(
    df["income"],
    df["high_income"]
)

income_check
```
